```python
#Question1
import pandas as pd
#Creat the original dataset
data = {
    "Student":["Ali","Sara","John"],
    "Math":[75,88,65],
    "Science":[80,91,70],
    "English":[72,85,68]
}

df = pd.DataFrame(data)

#Display the original dataset
print("Original Dataset:")
print(df)

#Reshape the dataset into tidy format
tidy_df = pd.melt(
    df,
    id_vars=["Student"],
    var_name="subject",
    value_name="Score"
)

print("\nTidy Dataset:")
print(tidy_df)

#Question2
import pandas as pd

#Creat the original dataset
data={
    "Country":["Malaysia","Thailand","Indonesia"],
    "Q1_Sales":[12000,10000,14000],
    "Q2_Sales":[15000,11000,12500],
    "Q3_Sales":[14000,13000,15000],
    "Q4_Sales":[18000,16000,17000]
}

df = pd.DataFrame(data)

#Display the original dataset
print("Original Dataset:")
print(df)

#Convert the dataset into tidy format
tidy_df = pd.melt(
    df,
    id_vars=["Country"],
    var_name="Quarter",
    value_name="Sales"
)

#Remove "_Sales" from the Quarter colume
tidy_df["Quarter"] = tidy_df["Quarter"].str.replace("_Sales","",regex=False)

print("\nTidy Dataset:")
print(tidy_df)

#Question3
import pandas as pd
import matplotlib.pyplot as plt

#Creat the dataset
data = {
    "Student": ["A","B","C","D","E","F"],
    "Study_Hours": [2,5,3,7,4,8],
    "Score": [55,72,60,85,68,90]
}

df = pd.DataFrame(data)

#Display the dataset
print("Dataset:")
print(df)

#1.What is the average study time?
print("Average study time:",df["Study_Hours"].mean())

#2.What is the average score?
print("Average score:",df["Score"].mean())

#3.Which student obtained the highest score?
highest_score = df.loc[df["Score"].idxmax()]
print("Student with the highest score:",highest_score["Student"])
print("Highest score:", highest_score["Score"])

#4.Which student studied the most?
most_study = df.loc[df["Study_Hours"].idxmax()]
print("Student who studied the most:",most_study["Student"])
print("Study hours:",most_study["Study_Hours"])

#5.Does studying more hours appear to be associated with a higher score?
print("Study hours and scores appear to have a positive relationship.")

#6.Create a scatter plot of Study_Hours vs Score
plt.scatter(df["Study_Hours"],df["Score"])
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score")
plt.show()
```
